```python
tables_to_check = ["customers", "geolocation", "sellers", "product_category_name_translation", "products"]
for t in tables_to_check:
    try:
        result = pd.read_sql(f"SELECT COUNT(*) as cnt FROM {t}", con=engine)
        print(f"{t}: {result['cnt'][0]} rows")
    except Exception as e:
        print(f"{t}: error - {e}")
```
